```python
from google.colab import drive
drive.mount('/content/drive')

import zipfile
import os
import shutil

# Step 2: Unzip (PEHLE)
zip_path = "/content/drive/MyDrive/pneumonia_projectfyp/pneumonia_data_final.zip"
extract_path = "/content/dataset"

os.makedirs(extract_path, exist_ok=True)
with zipfile.ZipFile(zip_path, 'r') as zip_ref:
    zip_ref.extractall(extract_path)

print("Step 2 done: Extracted.")

# Step 3: Fix Windows-style backslash filenames (BAAD MEIN)
flat_dir = "/content/dataset"
fixed_dir = "/content/dataset_fixed"

count = 0
for filename in os.listdir(flat_dir):
    full_path = os.path.join(flat_dir, filename)
    if os.path.isfile(full_path) and "\\" in filename:
        parts = filename.split("\\")
        new_path = os.path.join(fixed_dir, *parts)
        os.makedirs(os.path.dirname(new_path), exist_ok=True)
        shutil.move(full_path, new_path)
        count += 1

print(f"Step 3 done: Fixed {count} files.")
print("\nNew structure:")
for root, dirs, files in os.walk(fixed_dir):
    level = root.replace(fixed_dir, "").count(os.sep)
    if level <= 2:
        print("  " * level + os.path.basename(root) + "/")
```

```python
detection_root = "/content/dataset_fixed/final"
cls_root = "/content/dataset_cls"

split_map = {"train": "train", "valid": "val", "test": "test"}

for src_split, dst_split in split_map.items():
    img_dir = f"{detection_root}/{src_split}/images"
    lbl_dir = f"{detection_root}/{src_split}/labels"

    normal_dir = f"{cls_root}/{dst_split}/NORMAL"
    pneumonia_dir = f"{cls_root}/{dst_split}/PNEUMONIA"
    os.makedirs(normal_dir, exist_ok=True)
    os.makedirs(pneumonia_dir, exist_ok=True)

    for fname in os.listdir(img_dir):
        img_path = os.path.join(img_dir, fname)
        label_path = os.path.join(lbl_dir, os.path.splitext(fname)[0] + ".txt")

        is_pneumonia = os.path.exists(label_path) and os.path.getsize(label_path) > 0

        dest_dir = pneumonia_dir if is_pneumonia else normal_dir
        shutil.copy(img_path, os.path.join(dest_dir, fname))

    n_normal = len(os.listdir(normal_dir))
    n_pneumonia = len(os.listdir(pneumonia_dir))
    print(f"{dst_split}: NORMAL={n_normal}, PNEUMONIA={n_pneumonia}")

print(f"\nClassification dataset ready at: {cls_root}")
```

```python
from ultralytics import YOLO
from sklearn.metrics import precision_recall_curve
import matplotlib.pyplot as plt
import os

# Load the 3-class nano model (same one used in Flask app)
model_path = "/content/drive/MyDrive/pneumonia_projectfyp/runs_cls/yolo26n_cls_3class/weights/best.pt"
yolo_model = YOLO(model_path)

y_true = []
y_scores = []  # confidence for PNEUMONIA class specifically

test_dir = f"{cls_root}/test"
for class_name in ["NORMAL", "PNEUMONIA"]:
    class_dir = f"{test_dir}/{class_name}"
    for fname in os.listdir(class_dir):
        img_path = os.path.join(class_dir, fname)
        result = yolo_model.predict(img_path, verbose=False)[0]
        pneumonia_idx = list(result.names.values()).index("PNEUMONIA")
        pneumonia_score = float(result.probs.data[pneumonia_idx])

        y_true.append(1 if class_name == "PNEUMONIA" else 0)
        y_scores.append(pneumonia_score)

# Compute precision-recall curve
precision, recall, thresholds = precision_recall_curve(y_true, y_scores)

# Plot it
plt.figure(figsize=(7, 6))
plt.plot(recall, precision, linewidth=2, color='#2563eb')
plt.xlabel('Recall')
plt.ylabel('Precision')
plt.title('Precision-Recall Curve (PNEUMONIA vs NORMAL)')
plt.grid(alpha=0.3)
plt.savefig('/content/precision_recall_curve.png', dpi=150, bbox_inches='tight')
plt.show()

print("Saved to /content/precision_recall_curve.png")
```

```python
from ultralytics import YOLO
from sklearn.metrics import confusion_matrix, classification_report, roc_auc_score
import os

# Load the 3-class nano model (same one used in your Flask app)
model_path = "/content/drive/MyDrive/pneumonia_projectfyp/runs_cls/yolo26n_cls_3class/weights/best.pt"
yolo_model = YOLO(model_path)

# Run predictions on your full test set (NORMAL + PNEUMONIA only, skipping OTHER)
y_true = []
y_pred = []
y_scores = []  # confidence for PNEUMONIA class specifically

test_dir = f"{cls_root}/test"
for class_name in ["NORMAL", "PNEUMONIA"]:
    class_dir = f"{test_dir}/{class_name}"
    for fname in os.listdir(class_dir):
        img_path = os.path.join(class_dir, fname)
        result = yolo_model.predict(img_path, verbose=False)[0]
        pred_label = result.names[result.probs.top1]
        pneumonia_idx = list(result.names.values()).index("PNEUMONIA")
        pneumonia_score = float(result.probs.data[pneumonia_idx])

        y_true.append(class_name)
        y_pred.append(pred_label)
        y_scores.append(pneumonia_score)

print(f"Total images evaluated: {len(y_true)}")

# Confusion matrix + full report
labels = ["NORMAL", "PNEUMONIA"]
cm = confusion_matrix(y_true, y_pred, labels=labels)
print("\nConfusion Matrix:")
print(f"           Predicted NORMAL  Predicted PNEUMONIA")
print(f"True NORMAL      {cm[0][0]:>5}              {cm[0][1]:>5}")
print(f"True PNEUMONIA   {cm[1][0]:>5}              {cm[1][1]:>5}")

print("\nFull Report:")
print(classification_report(y_true, y_pred, labels=labels))

# ROC-AUC
y_true_binary = [1 if x == "PNEUMONIA" else 0 for x in y_true]
auc = roc_auc_score(y_true_binary, y_scores)
print(f"ROC-AUC: {auc:.4f}")

# Threshold sweep
print("\nThreshold sweep:")
for threshold in [0.3, 0.4, 0.5, 0.6, 0.7, 0.8]:
    y_pred_at_threshold = ["PNEUMONIA" if s >= threshold else "NORMAL" for s in y_scores]
    cm_t = confusion_matrix(y_true, y_pred_at_threshold, labels=labels)
    tn, fp, fn, tp = cm_t.ravel()
    precision = tp/(tp+fp+1e-9)
    recall = tp/(tp+fn+1e-9)
    print(f"Threshold {threshold}: FP={fp}, FN={fn}, Precision={precision:.3f}, Recall={recall:.3f}")

# Find and print each false positive individually
print("\n--- False Positives (NORMAL predicted as PNEUMONIA) ---")
fp_count = 0
for i, (true, pred, score) in enumerate(zip(y_true, y_pred, y_scores)):
    if true == "NORMAL" and pred == "PNEUMONIA":
        fp_count += 1
        print(f"FP #{fp_count}: True=NORMAL, Predicted=PNEUMONIA, Confidence={score:.3f}")

print(f"\nTotal false positives: {fp_count}")
```

```python
MODEL_PATHS = {
    "YOLO-Nano":   "/content/drive/MyDrive/pneumonia_projectfyp/runs_cls/yolo26n_cls/weights/best.pt",
    "YOLO-Small":  "/content/drive/MyDrive/pneumonia_projectfyp/runs_cls/yolo26s_cls/weights/best.pt",
    "YOLO-Medium": "/content/drive/MyDrive/pneumonia_projectfyp/runs_cls/yolo26m_cls/weights/best.pt",
}

# Sanity check first
for name, path in MODEL_PATHS.items():
    print(f"{name}: {'FOUND' if os.path.isfile(path) else 'MISSING'} -> {path}")
```

```python
results_summary = {}

print("Evaluating all three models on the SAME test set...\n")

for name, path in MODEL_PATHS.items():
    print(f"--- Evaluating {name} ---")
    model = YOLO(path)
    metrics = model.val(data=cls_root, split="test", verbose=False)

    results_summary[name] = {
        "top1_accuracy": float(metrics.top1),
        "top5_accuracy": float(metrics.top5),
        "params_millions": sum(p.numel() for p in model.model.parameters()) / 1e6,
        "speed_ms": metrics.speed.get("inference", 0.0),
    }
    print(f"{name}: top1={metrics.top1:.4f}, params={results_summary[name]['params_millions']:.1f}M, "
          f"speed={results_summary[name]['speed_ms']:.2f}ms\n")

print("\n" + "=" * 65)
print(f"{'Model':<15}{'Top-1 Acc':<12}{'Top-5 Acc':<12}{'Params(M)':<12}{'Speed(ms)':<10}")
print("=" * 65)
for name, r in results_summary.items():
    print(f"{name:<15}{r['top1_accuracy']:<12.4f}{r['top5_accuracy']:<12.4f}"
          f"{r['params_millions']:<12.1f}{r['speed_ms']:<10.2f}")
print("=" * 65)

TIE_THRESHOLD = 0.001
best_name = max(results_summary, key=lambda k: results_summary[k]["top1_accuracy"])
best_acc = results_summary[best_name]["top1_accuracy"]
tied_models = [name for name, r in results_summary.items() if abs(r["top1_accuracy"] - best_acc) <= TIE_THRESHOLD]

if len(tied_models) > 1:
    best_name = min(tied_models, key=lambda k: results_summary[k]["params_millions"])
    print(f"\nNote: {', '.join(tied_models)} were within {TIE_THRESHOLD:.1%} — picking most efficient.")

print("\n" + "=" * 65)
print(f"BEST MODEL: {best_name}")
print("=" * 65)
```

```python
!pip install -q grad-cam

import cv2
import numpy as np
import torch
import torch.nn as nn
from pytorch_grad_cam import EigenCAM
from pytorch_grad_cam.utils.image import show_cam_on_image
import matplotlib.pyplot as plt
import os

# Load model
best_model_path = MODEL_PATHS["YOLO-Small"]
yolo_model = YOLO(best_model_path)
raw_model = yolo_model.model
raw_model.eval()

class YOLOWrapper(nn.Module):
    def __init__(self, model):
        super().__init__()
        self.model = model
    def forward(self, x):
        outputs = self.model(x)
        if isinstance(outputs, (tuple, list)):
            outputs = outputs[0]
        return outputs

cam_model = YOLOWrapper(raw_model)
cam_model.eval()
if torch.cuda.is_available():
    cam_model = cam_model.cuda()

target_layers = [raw_model.model[-2]]

class SumTarget:
    def __call__(self, model_output):
        return model_output.sum()

cam = EigenCAM(cam_model, target_layers)

# Pick test image
test_dir = f"{cls_root}/test/PNEUMONIA"
sample_img = os.listdir(test_dir)[0]
img_path = os.path.join(test_dir, sample_img)

img = cv2.imread(img_path)
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
img = cv2.resize(img, (416, 416))
rgb_img = img.astype(np.float32) / 255.0
input_tensor = torch.from_numpy(rgb_img).permute(2, 0, 1).unsqueeze(0).float()
if torch.cuda.is_available():
    input_tensor = input_tensor.cuda()

# Predict
results = yolo_model.predict(img_path, verbose=False)
result = results[0]
predicted_class = result.names[result.probs.top1]
confidence = result.probs.top1conf.item()
print(f"Prediction: {predicted_class} ({confidence:.2%})")

# Grad-CAM
grayscale_cam = cam(input_tensor, targets=[SumTarget()])[0, :, :]
cam_image = show_cam_on_image(rgb_img, grayscale_cam, use_rgb=True)

fig, axes = plt.subplots(1, 2, figsize=(12, 6))
axes[0].imshow(rgb_img)
axes[0].set_title("Original X-Ray")
axes[0].axis('off')
axes[1].imshow(cam_image)
axes[1].set_title(f"Grad-CAM: {predicted_class} ({confidence:.1%})")
axes[1].axis('off')
plt.tight_layout()
plt.savefig("/content/gradcam_small_cls.png", dpi=150)
plt.show()
```

```python
import torchvision
import os
import random

print("Downloading generic image source (CIFAR-10)...")
cifar = torchvision.datasets.CIFAR10(root="/content/cifar_source", download=True)

for split in ["train", "val", "test"]:
    os.makedirs(f"{cls_root}/{split}/OTHER", exist_ok=True)

counts = {"train": 3000, "val": 380, "test": 380}

indices = list(range(len(cifar)))
random.seed(42)
random.shuffle(indices)

idx_pointer = 0
for split, n in counts.items():
    for i in range(n):
        img, _ = cifar[indices[idx_pointer]]
        img = img.resize((416, 416))
        img.save(f"{cls_root}/{split}/OTHER/other_{idx_pointer}.jpg")
        idx_pointer += 1
    print(f"{split}: added {n} OTHER images")

print("\nFinal dataset structure:")
for split in ["train", "val", "test"]:
    for cls in ["NORMAL", "PNEUMONIA", "OTHER"]:
        path = f"{cls_root}/{split}/{cls}"
        if os.path.isdir(path):
            print(f"  {split}/{cls}: {len(os.listdir(path))} images")
```
